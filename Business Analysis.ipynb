## 2. Visualizations

### Pie Plot of Churn

```python
churn = df['churn'].value_counts().reset_index()
churn
```

```python
plt.figure(figsize= (8,6))
plt.pie(
    x=churn['count'],
    labels = churn['churn'],
    autopct = "%1.1f%%",
    startangle = 90
)
plt.show()
```

```python
df['tenure'].value_counts().reset_index()
```

```python
df.head()
```

```python
tenure_bins = [-1, 12, 36, df['tenure'].max()]
tenure_labels = ['1-12 Months', '13-36 Months', '37+ Months']

df['tenure_category'] = pd.cut(df['tenure'], bins = tenure_bins, labels = tenure_labels)
```

```python
tenure_cat = df['tenure_category'].value_counts().reset_index()
tenure_cat.rename(columns = {'tenure_category': "Category", 'count': "Count"}, inplace = True)
```

```python
tenure_cat
```

```python
plt.figure(figsize = (8, 6))
plt.pie(
    x= 'Count',
    labels = 'Category',
    data = tenure_cat,
    autopct = '%1.1f%%',
    startangle = 90
)
plt.title("Pie Chart showing the Distribution of Tenure Categories")
plt.show()
```
